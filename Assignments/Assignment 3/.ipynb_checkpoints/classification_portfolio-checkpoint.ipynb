## 1. Introduction
The objective of this project is to classify suicide risk levels based on demographic and socio-economic factors using the `master.csv` dataset.
We will preprocess the data, perform Exploratory Data Analysis (EDA), and build both a Neural Network and Classical Machine Learning models.
**Flow**: Data Cleaning -> Visualization -> Model Building -> Hyperparameter Optimization -> Feature Selection -> Final Model.

## 2. Methodology
1.  **Data Preprocessing**: Handling missing values, encoding, and scaling.
2.  **EDA**: Visualizing distributions and correlations.
3.  **Task 1 - Neural Network**: `MLPClassifier` (Scikit-Learn).
4.  **Task 2 - Classical ML Models**: 
    -   **Models**: Logistic Regression (Poly+Balanced) and KNN.
    -   **Optimization**: GridSearchCV.
    -   **Feature Selection**: RFE (for LR) and SelectKBest (for KNN).
    -   **Final Model**: Built on selected features with optimal parameters.
5.  **Evaluation**: Accuracy, Precision, Recall, F1-Score.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler, PolynomialFeatures
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.feature_selection import RFE, SelectKBest, f_classif
from sklearn.pipeline import Pipeline

import warnings
warnings.filterwarnings('ignore')

print("Libraries Imported Successfully")
```

## 3. Data Loading and Preprocessing

```python
df = pd.read_csv('master.csv')
df = df.drop(['HDI for year', 'country-year'], axis=1)
df.columns = df.columns.str.strip()
df.dropna(inplace=True)

# Create Target Class
bins = [-1, 10, 30, 1000]
labels = ['Low', 'Medium', 'High']
df['Suicide_Risk'] = pd.cut(df['suicides/100k pop'], bins=bins, labels=labels)
df = df.drop(['suicides/100k pop', 'suicides_no'], axis=1)

# Encode Categoricals (except country)
le = LabelEncoder()
cat_cols = ['sex', 'age', 'generation']  # removed 'country'
for col in cat_cols:
    df[col] = le.fit_transform(df[col])

target_map = {'Low': 0, 'Medium': 1, 'High': 2}
df['Suicide_Risk'] = df['Suicide_Risk'].map(target_map)

# Clean GDP
df['gdp_for_year ($)'] = df['gdp_for_year ($)'].astype(str).str.replace(',', '').astype(float)

# Log Transforms
df['log_gdp'] = np.log1p(df['gdp_for_year ($)'])
df['log_pop'] = np.log1p(df['population'])
df['log_gdp_per_capita'] = np.log1p(df['gdp_per_capita ($)'])
df = df.drop(['gdp_for_year ($)', 'population', 'gdp_per_capita ($)'], axis=1)

print("Data Preprocessed. Shape:", df.shape)
df.head()
```

## 4. Exploratory Data Analysis (EDA)

```python
# 1. Distribution of Target Variable
plt.figure(figsize=(8, 6))
sns.countplot(x='Suicide_Risk', data=df, palette='viridis')
plt.title('Distribution of Suicide Risk Levels (0=Low, 1=Medium, 2=High)')
plt.xlabel('Risk Level')
plt.show()

# 2. Correlation Matrix
plt.figure(figsize=(10, 8))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Feature Correlation Matrix')
plt.show()
```

```python

# 3. Box Plots for Numerical Features vs Risk
num_cols = ['log_gdp', 'log_pop', 'log_gdp_per_capita']
plt.figure(figsize=(15, 5))
for i, col in enumerate(num_cols):
    plt.subplot(1, 3, i+1)
    sns.boxplot(x='Suicide_Risk', y=col, data=df, palette='Set2')
    plt.title(f'{col} vs Suicide Risk')
plt.tight_layout()
plt.show()
```

```python

# 4. Impact of Age and Sex on Suicide Risk
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.countplot(x='age', hue='Suicide_Risk', data=df, palette='pastel')
plt.title('Suicide Risk by Age Group (Encoded)')
plt.legend(title='Risk Level')

plt.subplot(1, 2, 2)
sns.countplot(x='sex', hue='Suicide_Risk', data=df, palette='pastel')
plt.title('Suicide Risk by Sex (Encoded)')
plt.legend(title='Risk Level')
plt.show()
```

```python

# Separate features and target
X = df.drop('Suicide_Risk', axis=1)
y = df['Suicide_Risk']

# Keep only numeric columns for scaling
X_numeric = X.select_dtypes(include=np.number)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_numeric, y, test_size=0.2, random_state=42)

# Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

```

```python

# Create Polynomial Features for Logistic Regression
poly = PolynomialFeatures(degree=2, include_bias=False)
X_train_poly = poly.fit_transform(X_train_scaled)
X_test_poly = poly.transform(X_test_scaled)
print("Polynomial Features Shape:", X_train_poly.shape)
```

## 4. Task 1: Neural Network (MLPClassifier)

```python

mlp = MLPClassifier(hidden_layer_sizes=(128, 64), activation='relu', solver='adam', max_iter=500, random_state=42)
mlp.fit(X_train_scaled, y_train)
y_pred_mlp = mlp.predict(X_test_scaled)
print("--- MLP Accuracy: {:.2f}% ---".format(accuracy_score(y_test, y_pred_mlp)*100))
```

## 5. Task 2: Classical Machine Learning Models

### 5.1 Hyperparameter Optimization

```python

# 1. Logistic Regression (using Polynomial Features)
lr = LogisticRegression(class_weight='balanced', max_iter=2000, solver='lbfgs')
param_grid_lr = {'C': [0.1, 1, 10]}
grid_lr = GridSearchCV(lr, param_grid_lr, cv=3, verbose=1, n_jobs=-1)
grid_lr.fit(X_train_poly, y_train)

best_lr = grid_lr.best_estimator_
print(f"Best LR Params: {grid_lr.best_params_}")
```

```python

# 2. KNN (using Standard Scaled Features)
knn = KNeighborsClassifier()
param_grid_knn = {'n_neighbors': [3, 5, 7, 9], 'weights': ['distance']}
grid_knn = GridSearchCV(knn, param_grid_knn, cv=3, verbose=1, n_jobs=-1)
grid_knn.fit(X_train_scaled, y_train)

best_knn = grid_knn.best_estimator_
print(f"Best KNN Params: {grid_knn.best_params_}")
```

### 5.2 Feature Selection

```python

# Feature Selection for Logistic Regression (RFE)
# Using RFE to select top 20 important polynomial features
rfe_lr = RFE(estimator=best_lr, n_features_to_select=20)
rfe_lr.fit(X_train_poly, y_train)

selected_features_mask = rfe_lr.support_
X_train_lr_sel = rfe_lr.transform(X_train_poly)
X_test_lr_sel = rfe_lr.transform(X_test_poly)

print(f"LR Features reduced from {X_train_poly.shape[1]} to {X_train_lr_sel.shape[1]}")
```

```python

# Feature Selection for KNN (SelectKBest)
# KNN doesn't work with RFE, so we use SelectKBest (ANOVA F-value)
selector_knn = SelectKBest(score_func=f_classif, k=10) # Select top 10 features
X_train_knn_sel = selector_knn.fit_transform(X_train_scaled, y_train)
X_test_knn_sel = selector_knn.transform(X_test_scaled)

print(f"KNN Features reduced from {X_train_scaled.shape[1]} to {X_train_knn_sel.shape[1]}")
```

### 5.3 Final Model & Evaluation

```python

# Training Final Models on Selected Features

# 1. Final Logistic Regression
final_lr = grid_lr.best_estimator_ # Re-use best params
final_lr.fit(X_train_lr_sel, y_train)
y_pred_final_lr = final_lr.predict(X_test_lr_sel)

# 2. Final KNN
final_knn = grid_knn.best_estimator_
final_knn.fit(X_train_knn_sel, y_train)
y_pred_final_knn = final_knn.predict(X_test_knn_sel)

# Evaluation
def print_results(y_true, y_pred, name):
    print(f"--- {name} Final Performance ---")
    print(classification_report(y_true, y_pred))
    return accuracy_score(y_true, y_pred)

acc_lr = print_results(y_test, y_pred_final_lr, "Logistic Regression (Selected Features)")
acc_knn = print_results(y_test, y_pred_final_knn, "KNN (Selected Features)")
```


## 6. Comparison and Conclusion
We optimized the models using GridSearch and refined them using Feature Selection.
Below is the final comparison.


```python

results = pd.DataFrame({
    'Model': ['MLP Neural Network', 'Final Logistic Regression', 'Final KNN'],
    'Accuracy': [accuracy_score(y_test, y_pred_mlp), acc_lr, acc_knn]
})
print(results)
sns.barplot(x='Model', y='Accuracy', data=results, palette='viridis')
plt.title("Final Model Accuracy Comparison")
plt.show()
```
